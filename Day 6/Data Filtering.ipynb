# Data Filtering in Pandas

```python
# Objective

# Create a filtered version of the dataset by:

# Filtering rows
# Selecting columns
# Sorting data
```

```python
# 1. Import Pandas
import pandas as pd
```

```python
# 2. Read CSV File
df = pd.read_csv("SalesDataset.csv")
```

```python
# 3. Display Dataset
df.head()
```

```python
# 4. Select Columns
df["Product"]
```

```python
df[["Product", "TotalSales", "City"]]
```

```python
# 5. Filter Rows
df[df["TotalSales"] > 10000]
```

```python
# Only Delhi records
df[df["City"] == "Delhi"]
```

```python
# Shoes category
df[df["Category"] == "Office"]
```

```python
# Quantity greater than 5
df[df["Discount"] > 10]
```

```python
# 6. Multiple Conditions
df[(df["City"] == "Delhi") & (df["TotalSales"] > 10000)]
```

```python
# Mumbai OR Jeans
df[(df["City"] == "Mumbai") | (df["Category"] == "Jeans")]
```

```python
# 7. isin()
df[df["City"].isin(["Delhi", "Mumbai"])]
```

```python
# 8. Between()
df[df["TotalSales"].between(5000,15000)]
```

```python
# 9. String Filtering
# Starts with A
df[df["CustomerName"].str.startswith("A")]
```

```python
# Contains "phone"
df[df["Product"].str.contains("phone", case=False)]
```

```python
# 10. Sort Values
# Sales ascending
df.sort_values("TotalSales")
```

```python
# Sales descending
df.sort_values("TotalSales", ascending=False)
```

```python
# Multiple columns
df.sort_values(["City","TotalSales"], ascending=[True,False])
```

```python
# Largest Sales
# Top 10 sales
df.nlargest(10,"TotalSales")
```

```python
# Lowest 10 sales
df.nsmallest(10,"TotalSales")
```

```python
# Save Filtered Dataset
filtered = df[df["TotalSales"] > 10000]

filtered.to_csv("Filtered_Data.csv", index=False)
```
