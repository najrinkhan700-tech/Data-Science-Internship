# concate & Merge data

```python
import pandas as pd
```

```python
# 1. Concatenate (DataFrames ko jodna)

# 👉 concat() ka use hota hai:

# Multiple DataFrames ko stack (upar-neeche ya side-by-side) jodne ke liye
# pd.concat([df1, df2])
```

```python
df1 = pd.DataFrame({
    "Name" : ["Ram" , "Ravi"],
    "Age" : [25,30]
})

df2 = pd.DataFrame({
    "Name" : ["Amit", "Sita"],
    "Age" : [28, 22]
})

result = pd.concat([df1 , df2], axis = 0)  # conact Vertical / row wise
# result = pd.concat([df1 , df2], axis = 1)  # concat Horizontal / column wise
result
```

```python
# Merge in pandas 
# merge on comman colimn like sql
df1 = pd.DataFrame({
    "emp_id" : [1,2,3],
    "Name" : ['Ram', 'Ravi', 'Amit']
})

df2 = pd.DataFrame({
    "emp_id" : [1,2,4],
    "Salary" : [30000, 40000, 50000]
})
```

```python
 pd.merge(df1, df2, on = "emp_id")  # inner join
```

```python
pd.merge(df1, df2, on = "emp_id", how = "left")   # left join
```

```python
pd.merge(df1, df2, on = "emp_id", how = "right")   # Right join
```

```python
pd.merge(df1, df2, on = "emp_id", how = "outer")   # Full join
```

```python

# Join Type	Result
# Inner	Only matching
# Left	Left + matching right
# Right	Right + matching left
# Outer	All data  # Right join
```

```python
pd.merge(df1, df2, how = "inner", left_on = "emp_id", right_on = "emp_id")   # dono table m alag data ho tab
```

```python
pd.merge(df1, df2, how='inner', left_on='emp_id', right_on='emp_id')  #dono table mai name alag ho tab
```
