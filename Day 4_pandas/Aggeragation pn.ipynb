```python
import pandas as pd
```

```python
data = {
    'Name': ['Ram', 'Ravi', 'Amit', 'Ram', 'Ravi'],
    'Dept': ['IT', 'HR', 'IT', 'HR', 'IT'],
    'Salary': [30000, 25000, 40000, 32000, 28000]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
df["Dept"].value_counts()  # frequency of values in month column
```

```python
df[df["Dept"] == "IT"].value_counts()  #  frequency of values in month column where month =1
```

```python
df.groupby("Dept")["Salary"].mean()  # dept wise avg salary
```

```python
df.groupby("Dept")["Name"].count() # count of employees per department
```

```python
df.groupby("Dept")["Salary"].agg(["sum", "mean", "min", "max"])  # multiple aggregations
```

```python
# multiple columns group by department + Name combination wise result 
df.groupby(["Dept", "Name"])["Salary"].sum()
```

```python
# Named aggeragation (Advanced)
df.groupby("Dept").agg(
    total_salary = ("Salary", "sum"),
    avg_salary = ("Salary" , "mean"),
    count_Salary = ("Salary", "count")
)
```
