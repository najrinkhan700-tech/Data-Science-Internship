# grouping

```python
import pandas as pd

# Employee Data Dictionary
data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    
    "Name": [
        "Aarav Sharma",
        "Priya Verma",
        "Rohit Singh",
        "Neha Khan",
        "Aditya Jain",
        "Simran Kaur",
        "Vikram Rao",
        "Pooja Das"
    ],
    
    "Department": [
        "IT",
        "HR",
        "Finance",
        "Marketing",
        "Sales",
        "IT",
        "HR",
        "Finance"
    ],
    
    "Age": [25, 28, 30, 24, 27, 26, 29, 31],
    
    "Salary": [45000, 40000, 55000, 38000, 47000, 52000, 41000, 60000],
    
    "Performance_Score": [88, 82, 91, 76, 85, 90, 80, 95]
}

# Create DataFrame
df = pd.DataFrame(data)

# Print DataFrame
print(df)
```

```python
groupd = df.groupby("Age")["Salary"].sum()
```

```python
groupd
```

```python
df.groupby("Department")["Salary"].mean()
```

```python
# group on multiple columns
groupd = df.groupby(["Age", "Name"])["Salary"].sum()
```
