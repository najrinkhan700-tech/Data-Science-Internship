# series

```python
s = pd.Series([10,20,30,40,50])
print(s)
print(s[0])
```

```python
# custom Index
s = pd.Series([100,200,300,400,500],index = ["a","b","c","d","e"])
print(s)
print(s.loc["b"])
```

```python
# Dictonary Series
data = {
    "Name" : "Ravi",
    "Course" : "Pyhton",
    "Fees" : 5000
}
s = pd.Series(data)
print(s)
print(s.values)
print(s.index)
print(s.dtype)
```

```python
# Difference: List vs Series
# List	Series
# Normal Python	Pandas Object
# No label	Index hota hai
# Less powerful	Data analysis ke liye useful

# DataFrame vs Series
# Series	DataFrame
# Single Column	Multiple Columns
# 1D	2D
```

# Data Frame

```python
# What is Dataframe?
# DataFrame Pandas ka ek 2-dimensional data structure hai jo data ko rows aur columns ke form mein 
#store karta hai,# bilkul Excel sheet ya database table ki tarah.

# DataFrame ek table hota hai jisme multiple rows aur columns hote hain.

# | Name   | Age | City    |
# | ------ | --- | ------- |
# | Aman   | 22  | Gwalior |
# | Rahul  | 24  | Indore  |
# | Simran | 21  | Bhopal  
```

```python
df = pd.DataFrame([10,11,12,13], columns = ["col_name"])
print(df)
print(type(df))
```

```python
# Using Dictonary of lists
data = {
    "Name" : ["Ram", "Ravi","John","Rony","Rohit","Sonu","Neha","Rupali","Sanjay","Monu"],
    "Age" : [22,23,21,20,22,25,23,22,21,19],
    "Salary" : [20000,24000,24000,23000,21000,34000,32000,31000,28000,30000]
}
df = pd.DataFrame(data)
print(df)
print(type(df))
```
