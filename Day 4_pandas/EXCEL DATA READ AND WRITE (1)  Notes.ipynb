```python
#3.c) Writing to Multiple Sheets

import pandas as pd

df1 = pd.DataFrame({'Name': ['Amit', 'Simran'], 'Marks': [85, 90]}) #first sheet table
df2 = pd.DataFrame({'Subject': ['Math', 'Science'], 'Teacher': ['Mr. A', 'Ms. B']})# second sheet table
  
with pd.ExcelWriter('First.xlsx') as writer: #we make a excel file in which we make sheets
    df1.to_excel(writer, sheet_name='Students', index=False)
    df2.to_excel(writer, sheet_name='Teachers', index=False)
    #we make the excel file called First and inside it we made 2 sheeys student and teachers 
```

```python
#Now we have to   read the created sheet1(Students) of First file(excel)
df = pd.read_excel('First.xlsx', sheet_name='Students')
print(df) 
```

```python
#Now we have to   read the created sheet2(Teachers) of First file(excel)
df = pd.read_excel('Abhisek.xlsx', sheet_name='Teachers')
print(df)
```
