```python

# 🧠 What is a Pivot Table?

# A pivot table means:

# “Turn your long data table into a summary table that shows information in a more organized and easy-to-read format.”

# It rearranges your data — just like you do in Excel Pivot Tables.




# | A (Company) | B (Branch) | C (Product) | D (Sales) |
# | ----------- | ---------- | ----------- | --------- |
# | foo         | one        | x           | 1         |
# | foo         | one        | y           | 3         |
# | foo         | two        | x           | 2         |
# | bar         | two        | y           | 5         |
# | bar         | one        | x           | 4         |
# | bar         | one        | y           | 1         |

# 🎯 Goal

# We want to see:

# “For each Company (A) and Branch (B), what are the sales of each Product (C)?”

# So we use:

# df.pivot_table(values='Dx', index=['A','B'], columns='C')


# 🔄 Step-by-Step Visualization
# Step 1 — Take A and B as rows
#     | A   | B   |
# | --- | --- |
# | bar | one |
# | bar | two |
# | foo | one |
# | foo | two |


# Step 2 — Take C as columns

# C has two values → x, y
# | A | B | x | y |
# | - | - | - | - |


# Step 3 — Fill in the D (Sales) values

# Now, fill each cell with the correct sales value from the table:

# | A   | B   | x   | y   |
# | --- | --- | --- | --- |
# | bar | one | 4   | 1   |
# | bar | two | NaN | 5   |
# | foo | one | 1   | 3   |
# | foo | two | 2   | NaN |
# ✅ If a combination doesn’t exist → NaN (missing)


# So this pivot table says:
# | Company | Branch | Product X | Product Y |
# | ------- | ------ | --------- | --------- |
# | bar     | one    | 4         | 1         |
# | bar     | two    | –         | 5         |
# | foo     | one    | 1         | 3         |
# | foo     | two    | 2         | –         |



# 📘 Summary
# Concept	Description
# index	What becomes the rows (like “Company”, “Branch”)
# columns	What becomes the columns (like “Product”)
# values	What fills the cells (like “Sales”)
# aggfunc	How to combine duplicates (default = mean)
```

```python
df.pivot_table(values=data,index=['A', 'B'],columns=['C'])
```
