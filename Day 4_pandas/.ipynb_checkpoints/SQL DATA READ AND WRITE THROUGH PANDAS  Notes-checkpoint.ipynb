```python

#.How to write data in database table(which is created in xammp)
```

```python
import pandas as pd
from sqlalchemy import create_engine

# Connection settings
username = 'root'
password = ''  # Default for XAMPP
host = 'localhost'
port = 3306
database = 'school record'
# Create engine
engine = create_engine(f"mysql+mysqlconnector://{username}:{password}@{host}:{port}/{database}") 
```

```python
data = {
    'id':[2,3,4],
    'name': ['Tina', 'Arun', 'Neha'],
    'marks': [87, 91, 76],
    
}

df = pd.DataFrame(data)
```

```python
df.to_sql(name='students record', con=engine, if_exists='append', index=False)
```

```python
data = {
    'id': ['7', '8', '9'],
    'name': ['Anu', 'Deepa', 'Rakesh'],
    'marks': [77, 69, 88],
    
}
df = pd.DataFrame(data)
df.to_sql('students record', con=engine, if_exists='append', index=False)
```

```python
#Q3. Replace entire table with new records
#Hint: Use if_exists='replace'
```

```python
df = pd.DataFrame({
    'id': ['10', '11'],
    'name': ['Kriti', 'Mohit'],
    'marks': [90, 92],
    
})
df.to_sql('students_record', con=engine, if_exists='replace', index=False)
```

```python
#Q4. Insert data and include index column


import pandas as pd
from sqlalchemy import create_engine

# Connection settings
username = 'root'
password = ''  # Default for XAMPP
host = 'localhost'
port = 3306
database = 'data'

# Create engine
engine = create_engine("mysql+pymysql://root:@localhost/data")

df = pd.DataFrame({
    'name': ['Simran', 'Amit'],
    'marks': [73, 85]
})
df.index.name = 'roll_no'
df.to_sql('raghavi', con=engine, if_exists='replace', index=True)
```

```python
#Q5. Read specific columns using SQL
import pandas as pd
from sqlalchemy import create_engine

# Connection settings
username = 'root'
password = ''  # Default for XAMPP
host = 'localhost'
port = 3306
database = 'data'

# Create engine
engine = create_engine("mysql+pymysql://root:@localhost/data") 

data = pd.read_sql('SELECT Name FROM raghavi', con=engine)
print(data)
```

```python
#Q6. Use index_label to rename index column in SQL table
import pandas as pd
from sqlalchemy import create_engine

# Connection settings
username = 'root'
password = ''  # Default for XAMPP
host = 'localhost'
port = 3306
database = 'data'

# Create engine
engine = create_engine("mysql+pymysql://root:@localhost/data")
df = pd.DataFrame({
    'name': ['Kiran', 'Alok'],
    'marks': [67, 77]
})
df.to_sql('raghavi', con=engine, if_exists='replace', index=True, index_label='id')
```

```python
# Q7. Write different DataFrames to different tables  (this automatically create tables in data Database
import pandas as pd
from sqlalchemy import create_engine

# Connection settings
username = 'root'
password = ''  # Default for XAMPP
host = 'localhost'
port = 3306
database = 'data'

# Create engine
engine = create_engine("mysql+pymysql://root:@localhost/data")

df1 = pd.DataFrame({'Name': ['Raj'], 'Marks': [91]})
df1.to_sql('s_class', con=engine, if_exists='replace', index=False)

df2 = pd.DataFrame({'Name': ['Neha'], 'Marks': [87]})
df2.to_sql('m_class', con=engine, if_exists='replace', index=False)
```

```python
# Q8. Use condition to filter data in SQL and load into Pandas
import pandas as pd
from sqlalchemy import create_engine
username = 'root'
password = ''  # Default for XAMPP
host = 'localhost'
port = 3306
database = 'data'

# Create engine
engine = create_engine("mysql+pymysql://root:@localhost/data")
query = "SELECT * FROM s_class WHERE Marks > 85"
high_scores = pd.read_sql(query, con=engine)
print(high_scores)
```

```python
#Q9. Try inserting empty DataFrame
import pandas as pd
from sqlalchemy import create_engine
username = 'root'
password = ''  # Default for XAMPP
host = 'localhost'
port = 3306
database = 'data'

# Create engine
engine = create_engine("mysql+pymysql://root:@localhost/data")
df_empty = pd.DataFrame(columns=['Name', 'Marks'])
df_empty.to_sql('empty_table', con=engine, if_exists='replace', index=False)
```

```python
#Q10. Insert timestamp column
import pandas as pd
from datetime import datetime
from sqlalchemy import create_engine
username = 'root'
password = ''  # Default for XAMPP
host = 'localhost'
port = 3306
database = 'data'

# Create engine
engine = create_engine("mysql+pymysql://root:@localhost/data")


df = pd.DataFrame({
    'Name': ['Varun'],
    'Marks': [88],
    'Updated': [datetime.now()]
})

df.to_sql('students_time', con=engine, if_exists='append', index=False)
```
