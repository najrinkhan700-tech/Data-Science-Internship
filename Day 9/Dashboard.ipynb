```python
# Step 1: Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Read Dataset
df = pd.read_csv("fashion_sales_1000_rows.csv")

df.head()
```

```python
# Step 2: Basic Business Analysis\
print("Total Sales :", df["Total Sales"].sum())
print("Average Sales :", df["Total Sales"].mean())
print("Maximum Sale :", df["Total Sales"].max())
print("Minimum Sale :", df["Total Sales"].min())
print("Total Orders :", df["Order ID"].count())
```

```python
# Step 3: Sales by Category (Bar Chart)
category_sales = df.groupby("Category")["Total Sales"].sum()

plt.figure(figsize=(8,5))
plt.bar(category_sales.index, category_sales.values)
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.savefig("bar_chart.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# Step 4: Daily Sales Trend (Line Chart)
daily_sales = df.groupby("Order Date")["Total Sales"].sum()

plt.figure(figsize=(10,5))
plt.plot(daily_sales.index, daily_sales.values, marker='o')
plt.title("Daily Sales Trend")
plt.xlabel("Order Date")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.savefig("line_chart.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# Step 5: Sales by Region (Pie Chart)
region_sales = df.groupby("Region")["Total Sales"].sum()

plt.figure(figsize=(7,7))
plt.pie(region_sales.values,
        labels=region_sales.index,
        autopct='%1.1f%%',
        startangle=90)

plt.title("Sales by Region")
plt.savefig("pie_chart.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# Step 6: KPI Summary Table
summary = pd.DataFrame({
    "Metric": [
        "Total Sales",
        "Average Sales",
        "Maximum Sale",
        "Minimum Sale",
        "Total Orders"
    ],
    "Value": [
        df["Total Sales"].sum(),
        df["Total Sales"].mean(),
        df["Total Sales"].max(),
        df["Total Sales"].min(),
        df["Order ID"].count()
    ]
})
plt.savefig("dashboard.png", dpi=300, bbox_inches="tight")
summary
```
