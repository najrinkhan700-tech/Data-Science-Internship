## Data Analysis using Pandas

# Step 1: Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Objective

# The objective of this project is to improve notebook formatting, code readability, comments, and visualizations using Pandas and Matplotlib.

# Step 2: Load Dataset

```python
# Load the dataset
df = pd.read_csv("SalesDataset.csv")

# Display first five rows
df.head()
```

# Step 3: Dataset Information

```python
# Display dataset information
df.info()
```

# Step 4: Data Cleaning

```python
# Check missing values
print(df.isnull().sum())

# Remove duplicate records
df = df.drop_duplicates()

# Fill missing values
df = df.fillna(0)
```

# Step 5: Data Analysis

```python
# Display statistical summary
df.describe()
```

```python
# Calculate Total Sales
total_sales = df["TotalSales"].sum()
print("Total Sales:", total_sales)
```

```python
# Calculate Average Sales
average_sales = df["TotalSales"].mean()
print("Average Sales:", average_sales)
```

```python
# Maximum Sales
maximum_sales = df["TotalSales"].max()
print("Maximum Sales:", maximum_sales)
```

```python
# Minimum Sales
minimum_sales = df["TotalSales"].min()
print("Minimum Sales:", minimum_sales)
```

# Step 6: Bar Chart

```python
category_sales = df.groupby("Category")["TotalSales"].sum()

plt.figure(figsize=(8,5))

category_sales.plot(kind="bar")

plt.title("Sales by Category")

plt.xlabel("Category")

plt.ylabel("Sales")

plt.grid(True)

plt.show()
```

# Step 7: Line Chart

```python
daily_sales = df.groupby("OrderDate")["TotalSales"].sum()

plt.figure(figsize=(10,5))

daily_sales.plot()

plt.title("Daily Sales Trend")

plt.xlabel("Date")

plt.ylabel("Sales")

plt.grid(True)

plt.show()
```

# Step 8: Pie Chart

```python
payment = df.groupby("PaymentMode")["TotalSales"].sum()

plt.figure(figsize=(7,7))

payment.plot(kind="pie", autopct="%1.1f%%")

plt.title("Payment Method Distribution")

plt.ylabel("")

plt.show()
```

## Business Insights

# - Total sales were calculated successfully.
# - Average sales value was identified.
# - Highest and lowest sales were analyzed.
# - Category-wise sales comparison was performed.
