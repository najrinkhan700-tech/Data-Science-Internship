```python
import pandas as pd
```

```python
# Read Dataset
df = pd.read_csv("SalesDataset.csv")
```

```python
# Clean the Dataset
# Remove duplicate rows
df = df.drop_duplicates()

# Remove rows with missing values
df = df.dropna()

# Convert Date column to datetime
df["Order_Date"] = pd.to_datetime(df["OrderDate"])

# Convert Total_Sales to numeric
df["Total_Sales"] = pd.to_numeric(df["TotalSales"])
```

```python
# Export Cleaned Dataset to CSV
df.to_csv("Cleaned_SalesDataset.csv", index=False)

print("Cleaned dataset exported successfully!")
```
