# Data Analysis

```python
# Task

# Calculate:

# Total → total sales/revenue
# Average → average sales
# Minimum → minimum sales
# Maximum → maximum sales
# Count → total number of records
```

```python
import pandas as pd
```

```python
df = pd.read_csv("fashion_sales_1000_rows.csv")
```

```python
df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
# Total Sales
total_sales = df["Total Sales"].sum()
print("Total Sales:", total_sales)
```

```python
# Average Sales
average_sales = df["Total Sales"].mean()
print("Average Sales:", average_sales)
```

```python
# Minimum Sales
minimum_sales = df["Total Sales"].min()
print("Minimum Sales:", minimum_sales)
```

```python
# Maximum Sales
maximum_sales = df["Total Sales"].max()
print("Maximum Sales:", maximum_sales)
```

```python
# Count
count_sales = df["Total Sales"].count()
print("Number of Sales Records:", count_sales)
```
