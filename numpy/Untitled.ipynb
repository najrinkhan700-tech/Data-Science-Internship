# Splitting

```python
# np.split()
# equal

# np.hsplit() - horizontally
# np.vsplit() - vertically it works only on where
# two or more dimension
```

```python
import numpy as np
```

```python
arr = np.array([10,20,30,40,50,60])
np.split(arr,2)
```

```python
print(np.hsplit(arr,3))
```

# Staking

```python
# np.vstack((arr1,arr2)) - vertically ,row wise
# np.hstack((arr1,arr2)) - horizontally , column wise
```

```python
import numpy as np
```

```python
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])
```

```python
# vstack()
np.vstack((arr1, arr2))
```

```python
# hstack()
np.hstack((arr1, arr2))
```
