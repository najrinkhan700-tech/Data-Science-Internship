# Reshaping

```python
import numpy as np
```

```python
arr = np.array([1,2,3,4,5,6])
```

```python
my_arr = arr.reshape(3,2)
my_arr
```

```python
my_arr = arr.reshape(2,3)
my_arr
```

```python
# flatten and ravel are used to when u want to convert the multidimensinal array
# in single dim
# .ravel() -> it return a view
# .flatten() -> it return a copy
```

```python
print(arr.flatten())
```

```python
arr = np.array([[1,2,3],[4,5,6]])
```

```python
print(arr.ravel())
```
