```python
# What is Broadcasting?
# Broadcasting is a feature that allows NumPy to perform operations on arrays of different shapes without manually copying data.
# Instead of creating a larger array, NumPy automatically expands the smaller array where possible
```

```python
import numpy as np
```

```python
# Adding a Number to an Array
arr = np.array([10, 20, 30, 40])
result = arr + 5
result
```

```python
# Adding Two Arrays
a = np.array([1,2,3])
b = np.array([10,20,30])
a + b
```

```python
# Broadcasting Row-wise
A = np.array([[1,2,3],
              [4,5,6]])
B = np.array([10,20,30])
A + B
```

```python
# Broadcasting Column-wise
A = np.array([[1,2,3],
              [4,5,6]])
B = np.array([[10],
              [20]])
A + B
```

```python
# Broadcasting Rules

# Two dimensions are compatible if:

# They are equal.
# One of them is 1
```

# What is Vectorization?

```python
# Vectorization means performing operations on an entire array at once instead of using Python loops.

# Vectorized code is:

# Faster
# Shorter
# Easier to read
```

```python
# Without Vectorization
arr = np.array([1,2,3,4])
result = []
for i in arr:
    result.append(i * 2)
print(result)
```

```python
# With Vectorization
arr = np.array([1,2,3,4])
arr * 2
```

```python
# Addition
a = np.array([1,2,3])
b = np.array([4,5,6])
a + b
```

```python
# Square Every Element
arr = np.array([1,2,3,4])

arr ** 2
```

```python
# Square Root
arr = np.array([4,9,16,25])
np.sqrt(arr)
```
