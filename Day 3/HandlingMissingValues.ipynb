# Handling Missing Values in NumPy

```python
# NumPy represents missing numeric values using:
# np.nan
```

```python
import numpy as np
```

```python
# Creating Missing Values
arr = np.array([10,20,np.nan,40,50])

arr
```

```python
# Check Missing Values
arr = np.array([10,20,np.nan,40])

np.isnan(arr)
```

```python
# Count Missing Values
arr = np.array([10,np.nan,20,np.nan,30])

np.sum(np.isnan(arr))
```

```python
# Remove Missing Values
arr = np.array([10,np.nan,20,30,np.nan])

clean = arr[~np.isnan(arr)]

clean
```

```python
# Replace Missing Values
arr = np.array([10,np.nan,20,np.nan])

arr[np.isnan(arr)] = 0
arr = np.nan_to_num(arr,nan = 100)  #default val ke liye
cleaned_arr2 = np.nan_to_num(arr, nan = 100)


arr
```

```python
# Replace with Mean
arr = np.array([10,np.nan,20,30])

mean = np.nanmean(arr)

arr[np.isnan(arr)] = mean

arr
```

```python
# Useful Functions for Missing Values
arr = np.array([10,np.nan,20,30])

print(np.nansum(arr))
print(np.nanmean(arr))
print(np.nanmax(arr))
```
