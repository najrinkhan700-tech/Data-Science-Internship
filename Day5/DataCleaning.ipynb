```python
# Data Cleaning Tasks: Handle missing values, remove duplicate records and correct data types.
```

```python
import pandas as pd
```

# read data set

```python
df = pd.read_csv("SalesDataset.csv")
```

```python
df
```

```python
# check missing values
df.isnull()
```

```python
df.isnull().sum()
```

```python
# remove missing values
df.dropna()
```

```python
# mean se
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

```python
df.isnull().sum()
```

```python
df["City"] = df["City"].fillna(df["City"].mode()[0])
```

```python
df.isnull().sum()
```

```python
df["UnitPrice"] = df["UnitPrice"].fillna(df["UnitPrice"].mean())
```

```python
df.isnull().sum()
```

# Handle duplicate data

```python
# check duplicate data
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

# Correct Data Types

```python
df.dtypes
```

```python
df["Age"] = df["Age"].astype(int)
```

```python
# save Clean dataset 
df.to_excel("Cleaned_Data.xlsx", index=False)
```
